```python
# Add nw product

new_product = input("Write your product info with sep (,) - ")
# 202    ,   Book    ,    200     ,   2  

# Cleaning Output

new_product = [item.strip() for item in new_product.split(",")]
new_product[0] = int(new_product[0])

ecommerce["Clothing"][new_product[0]] = {"name" : new_product[1],
                                         "price" : int(new_product[2]),
                                         "stock"  : int(new_product[3])}
ecommerce
```
