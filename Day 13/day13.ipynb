```python
birthday = input("Hi when your birthday comes?? DD-MM-YYYY - ")
birth_date = datetime.strptime(birthday,"%d-%m-%Y")

age = now.year - birth_date.year

print(f"You are {age} years old")
```

```python
birthday = input("When will be your next Birthday -")
birth_date = datetime.strptime(birthday,"%d-%m-%Y")

days_left = abs(now - birth_date)
print(f"Wow only {days_left.days} days left in your NEXT BIRTHDAY")
```

```python
import csv
from datetime import datetime
with open('date.csv','r') as file:
    reader = csv.DictReader(file)
    with open('formated_date.csv','w',newline='') as files:
        writer = csv.writer(files)
        writer.writerow(['Date','Day','Month','Year'])
        for r in reader:
            date_formated = datetime.strptime(r['date'],"%d-%m-%Y").date()
            day_name = date_formated.strftime("%A")
            month_name = date_formated.strftime("%B")
            year = date_formated.year
            writer.writerow([date_formated, day_name, month_name, year])
print("File created successfully")
```
