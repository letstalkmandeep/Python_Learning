```python
City = "Delhi"
Weather = 42

print(f"Delhi : {City} \nWeather : {Weather}")
```

```python
# Usinf .format() method
City = "Delhi"
Weather = 42

print("Delhi : {} \nWeather : {}".format(City,Weather))
```
