```python
# Write a program that takes an integer input from the user and checks whether the number is even or odd.

user = int(input("Write your number -").strip())

if (user % 2) == 0 :
    print(f"Number {user} is an even number")
else : print(f"Number {user} is odd number")
```

```python
# Write a program that asks for the user's age and checks if the person is eligible to vote (age 18 or older).

age = int(input("Write your age").strip())

if age >= 18 :
    print("You are eligible for giving vote")
else : print("You are not eligible for giving vote")
```

```python
# Write a program that takes an integer from the user and prints whether the number is positive, negative, or zero.

user = int(input("Write your number").strip())

if user < 0 :
    print(f"Your number {user} is negative")
elif user > 0 :
    print(f"Your number {user} is positive")
else : print("Your number is 0")
```

Write a program that takes a student's score (0-100) and prints their grade based on the following rules:

Score >= 90: Grade A
Score >= 75: Grade B
Score >= 50: Grade C
Score < 50: Grade F

```python
score = int(input("Write your score").strip())
if score >= 90:
    print("Grade A score")
elif score >= 75 :
    print("Grade B score")
elif score >=50 :
    print("Grade C score")
else : print("Grade F")
```

```python
# Write a program that takes three numbers as input and prints the largest of the three.

number = input("Write your 3 numbers use seperator (,)")
cleaned = [int(item.strip()) for item in number.split(",")]
max_number = max(cleaned)
print(f"Largest number from all of the values is {max_number}")
```

Write a program that asks the user for a year and checks whether it's a leap year or not. A year is a leap year if:

It is divisible by 4
But not divisible by 100 unless it is also divisible by 400

```python
year = int(input("Write your year").strip())

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0) :
    print(f"Year - {year} is a Leap Year")
else : print(f"Year - {year} is a normal year")
```
