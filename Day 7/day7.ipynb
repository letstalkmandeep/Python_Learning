```python
import random
rand = random.randint(1,100)
attempt = 5

while attempt > 0:
    number = int(input("What is your number pal?"))
    if number == rand:
        print(f"You won the number was {rand}")
        break
    elif number < rand:
        print(f"Your number is a bit lower\nOnly {attempt-1} left")
    elif number > rand:
        print(f"Your number is a bit higher\nOnly {attempt-1} left")

    attempt -= 1

    if attempt == 0:
        print(f"Your lose the random generated was {rand}")
```
