```python
# Write a program that prints numbers from 1 to 10 using a for loop.

for i in range(1,10+1):
    print(i)
```

```python
# Write a program that prints numbers from 1 to 10 using a while loop.

num = 1
while num <= 10:
    print(num)
    num += 1
```

```python
# Write a program to calculate the sum of the first 10 natural numbers using a for loop.

sum = 0
for i in range(1,10+1):
    print(i)
    sum = sum + i
print(sum)
```

```python
#Write a guessing game where the user has to guess a secret number between 1 and 10 using a while loop. The program should keep asking until they guess correctly
import random
win = random.randint(1,100)
attempt = 0
print("Welcome to the Game guess the number man")
while attempt >= 0 :
    number = int(input("Write your number between 1 to 100"))
    if number == win:
        print(f"You won number was {win}\nTotal attempts for winning this were {attempt}")
        break
    elif number > win:
        print("Your Number is a bit higher")
    else :
        print("Your Number is a bit lower")
    attempt += 1
```
