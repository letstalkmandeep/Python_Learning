# Day 10 — Python File Handling Practice

Practice reading, writing, and appending files. Run cells from this notebook's folder so the relative file paths work. The examples use the files in this practice folder.

## 1. Read a text file
Use `with open(...)` so Python closes the file automatically.

```python
with open('example.txt', 'r', encoding='utf-8') as file:
    content = file.read()

print(content)
```

## 2. Read line by line

```python
with open('example2.txt', 'r', encoding='utf-8') as file:
    for line in file:
        print(line.rstrip())
```

## 3. Write to a file
Mode `w` creates a file or replaces its existing contents. This example writes to `practice_output.txt` so the original sample files remain unchanged.

```python
with open('practice_output.txt', 'w', encoding='utf-8') as file:
    file.write('First line created with write mode.\n')
    file.write('Second line created with write mode.\n')
```

## 4. Append to a file
Mode `a` adds content to the end instead of replacing what is already there.

```python
with open('practice_output.txt', 'a', encoding='utf-8') as file:
    file.write('This line was added using append mode.\n')

with open('practice_output.txt', 'r', encoding='utf-8') as file:
    print(file.read())
```

## 5. Read JSON data
The built-in `json` module parses a JSON file into Python dictionaries and lists.

```python
import json

with open('data.json', 'r', encoding='utf-8') as file:
    data = json.load(file)

print(data['project'])
for student in data['students']:
    print(student['name'], student['score'])
```

## 6. Read CSV data
The built-in `csv` module is enough for simple CSV practice; you do not need pandas for this example.

```python
import csv

with open('employees.csv', 'r', newline='', encoding='utf-8') as file:
    reader = csv.DictReader(file)
    for employee in reader:
        print(employee['first_name'], employee['department'], employee['salary_inr'])
```
