## Read, Append and Write CSV files

```python
import csv

with open('employees.csv','r') as file:
    reader = csv.reader(file)
    for r in reader:
        print(r)
```

```python
## Write a csv file
import csv
with open('employees.csv','w',newline='') as file :
    write = csv.writer(file)
    write.writerow(['Name','Salary','ID'])
    write.writerow(['Hussain',20000,'CD210'])
    write.writerow(['Mandeep',2000000,'CD1000'])
```

```python
print(open('employees.csv','r').read())
```

```python
with open('employees.csv','a',newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['SHolla',100000,'CD100'])
    writer.writerow(['My Name',1000,'CFA100'])
```

```python
with open('employees.csv','r') as file:
    reader = csv.reader(file)
    for r in reader:
        print(r)
```
