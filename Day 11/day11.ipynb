```python
# TRY with devide by zero error

try :
    x = 10/0
except ZeroDivisionError:
    print('Sorry you cannot divide a number by zero')
else:
    x
```

```python
# File not found error

try:
    with open('tesst.text','r') as file:
        print(file.read())
except FileNotFoundError:
    print('Your file do not exists')
```

```python
# Using Finally

try:
    10/0
except ZeroDivisionError:
    print('You are dividing this with zero')
else :
    10/2
finally:
    print(10/2)
```

```python
# IOError

try:
    with open('test.text','x') as file:
        print(file.read())
except FileNotFoundError:
    print('Your file not exists error')
except IOError:
    print('Error while opening the file')
```
