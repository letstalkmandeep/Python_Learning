```python
class OutofRangeError(Exception):
    pass

def checknumber(x):
    if x < 1 or x > 100:
        raise OutofRangeError('Number must be between 1 to 100')
    else:
        print(x)

try:
    num = int(input('Write your number :'))
    checknumber(num)
except OutofRangeError as e:
    print(e)
```

```python
try:
    age = int(input('What is your AGE'))
    print(age)
except ValueError:
    if age < 1 or age > 120:
        print('Your age is not in range 1 to 120')
```

```python
def age(num):
    try :
        num = int(num)
        if num < 1 or num > 120:
            raise ValueError(f"Your age - {num} is not in range 1 to 120")
        return num
    except ValueError as e:
        print(e)
```

```python
age(1000)
```

```python
class GameOverError(Exception):
    pass

def simulation(lives):
    lives = int(lives)
    if lives <= 0:
        print('Write a valid life')
    else :
        print(f"Your total lifes are {lives}")
        try :
            while lives > 0:
                lives -= 1
                if lives == 0:
                    raise GameOverError('Your Game is Over Life remains 0')
                else:
                    print(f"Only {lives} lifes remains")
        except GameOverError as e:
            print(e)
```

```python
simulation(5)
```

```python
class UnderAgeError(Exception):
    pass

def check_voter_eligibility(age):
    age = int(age)
    try:
        if age < 18:
            raise UnderAgeError(f'Your age - {age} is not greater than or equals to 18')
        else :
            print(f'Your are under eligibility criteria  AGE - {age}')
    except UnderAgeError as e:
        print(e)
```

```python
check_voter_eligibility(18)
```
