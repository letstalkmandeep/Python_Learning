### Lambda Function

```python
# Write a lambda function that takes a number and returns its cube.

cube = lambda x : x**2
```

```python
cube(16)
```

```python
first_letter = lambda x : x[0]
```

```python
first_letter('hello')
```

```python
use = lambda x : x > 10
```

```python
use(20)
```

```python
last_letter = lambda x : x[-1]
```

```python
last_letter('Ol hohe')
```

```python
summiton = lambda x : sum(x)
```

```python
summiton([2,3,5,6])
```

```python
celcuis = [0,4,20]
celcuis
```

```python
[(i*9/5)+32 for i in celcuis]
```

```python
z = [2,5,6,8]
list(map(lambda x : x**2,z))
```

```python
k = ['Happy', 'Munna', 'Kanta', 'Nanna Patekar']
```

```python
list(map(lambda x: 'Hello ' + x, k))
```

```python
list(map(lambda x: len(x), k))
```

```python
list(map(lambda x : x+100,z))
```

```python
[i**2 for i in range(1,11)]
```

```python
num = [1,2,5,8,98,24,13]
[item for item in num if item%2 !=0 ]
```

```python
name = ['Suraj','Modi', 'Holly','Jonny']
[item.upper() for item in name]
```

```python
[item[0] for item in name]
```

```python
[i for i in range(1,21) if i%4 == 0]
```
