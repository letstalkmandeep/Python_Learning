```python
# Task 1: Write a function sum_of_list(lst) that takes a list of numbers as input and returns the sum of all the elements in the list.

def sum_of_list(numbers):
    summition = sum(numbers)
    return print(f"Summition of the list {numbers} is {summition}")
```

```python
num = input("Write your number with seperator ',' :").strip()
numbers = [int(item.strip()) for item in num.split(",")]
sum_of_list(numbers)
```

```python
# Task 2: Create a function multiply_list_elements(lst, multiplier)
# that multiplies each element in a list by a given multiplier and returns the new list.

def multiply_list_element(list,multiplier):
    mul = [item * multiplier for item in list]
    return print(f"Your list {list} * {multiplier} = {mul}")
```

```python
num = input("Write your numbers your wants to be multiply use seperator ',' :").strip()
multiplier = int(input("Write your multiplier :").strip())


list = [int(item.strip()) for item in num.split(",")]
multiply_list_element(list,multiplier)
```

```python
# Define a function string_length(str_list) that takes a list of strings 
# and returns a new list containing the length of each string.

def string_length(str_list):
    length = [len(item.strip()) for item in str_list]
    return length
```

```python
str_list = ["Mandeep ","OLAA ","OK  "]
string_length(str_list)
```

```python
# Create a function remove_duplicates(lst) that removes any duplicate elements from a list
#  and returns the list with unique elements only.

def remove_duplicates(dupe): 
    unique = list(set(dupe))
    return unique
```

```python
remove_duplicates([2,2,3,3,4,5,6])
```

```python
# Define a function calculate_discount(price, discount_percentage) that calculates and 
# returns the discounted price of an item based on the original price 
# and discount percentage.

def calculate_discount(price,discount_percentage):
    return price - (price*discount_percentage)
```

```python
calculate_discount(1000,0.1)
```
