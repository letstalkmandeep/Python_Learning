```python
def multiple(*x):
    mul = 1

    for i in x:
        mul *= i

    return mul
```

```python
multiple(1,2,3,4,5)
```

```python
def cal(numbers):
    total_sum = sum(numbers)
    avger = sum(numbers)/len(numbers)
    maximum = max(numbers)
    return print(f"Sum = {total_sum}\nAverage = {avger}\nMaximum = {maximum}")
```

```python
cal([1,2,3,4])
```

```python
def greet_time_based(name):
    from datetime import datetime
    hour = datetime.now().hour

    if hour >= 4 and hour < 12 :
        greetings = "Good Morning"
    elif hour >= 12 and hour < 16 :
        greetings = "Good Afternoon"
    elif hour >= 12 and hour < 20 :
        greetings = "Good Evening"
    else :
        greetings = "Good Night"
    return print(f"{greetings} {name} its {hour}o'clock")
```

```python
greet_time_based("Modi")
```
