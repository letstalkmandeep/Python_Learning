```python
# Create a variable pi with the value 3.14159. Use it to calculate and print the area of a circle with radius 7.

pi = 3.14159
print("Do you want to calculate the radius of the circle?")
radius = float(input("Got it what is the radius of your circle :"))
print(f"Area of your circle is {pi*(radius**2)}")
```
