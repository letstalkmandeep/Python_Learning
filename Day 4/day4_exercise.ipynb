```python
# Remove duplicates using a set
numbers = [1, 2, 3, 2, 4, 5, 1, 3, 6, 5]
numbers_set = set(numbers)
numbers = list(numbers_set)
numbers
```

```python
#List → Tuple → Set

numbers = [1, 2, 3, 2, 4, 5, 1, 3, 6, 5]
numbers_tuple = tuple(numbers)
numbers_set = set(numbers_tuple)
numbers = list(numbers_set)
numbers
```
